# Project Overview: Vegetation Health Classification using 1D-CNN

This notebook demonstrates a complete machine learning workflow for classifying vegetation health using multi-spectral satellite data (Sentinel-2). The approach treats spectral bands and indices as a 1D signal to capture local patterns.

## 1. Data Preparation
- **Source:** CSV data exported from Google Earth Engine (GEE).
- **Features (16 total):**
    - Spectral Bands: B2, B3, B4, B5, B6, B7, B8, B8A, B11, B12.
    - Vegetation Indices: NDVI, EVI, NDMI, SAVI, NDRE, BSI.
- **Preprocessing:**
    - Handling missing values and removing unnecessary columns (`.geo`, `system:index`).
    - Label shifting: Classes 1-4 shifted to 0-3 for PyTorch compatibility.
    - Normalization: `StandardScaler` applied to features.

## 2. Model Architecture: VegHealthCNN
- **Type:** 1D Convolutional Neural Network.
- **Layers:**
    - Three Conv1D blocks with BatchNorm, ReLU, and Dropout.
    - Global Average Pooling to reduce spatial dimensions.
    - Fully connected classifier with Dropout (0.4 and 0.3) to prevent overfitting.
- **Parameters:** ~10,243 trainable parameters.

## 4. Geospatial Inference & Visualization

```python
import numpy as np
import rasterio

# Path to your stacked indices
stacked_path = '/content/drive/MyDrive/NEURAL NETWORK/data/PHMA_VEG_2024_DL.tif'
output_rx_path = '/content/drive/MyDrive/NEURAL NETWORK/data/Nitrogen_Prescription_Full.tif'

with rasterio.open(stacked_path) as src:
    img = src.read()
    # Using nan_to_num to prevent propagation of NaN values through the formula
    bsi  = np.nan_to_num(img[10], nan=1.0)
    evi  = np.nan_to_num(img[11], nan=0.0)
    ndmi = np.nan_to_num(img[12], nan=0.0)
    ndre = np.nan_to_num(img[13], nan=0.0)
    ndvi = np.nan_to_num(img[14], nan=0.0)
    savi = np.nan_to_num(img[15], nan=0.0)

    meta = src.meta.copy()

    base_rate = 120.0
    veg_index = (ndvi + evi + savi) / 3

    # Need for Nitrogen (Lower NDRE = Higher Need)
    n_need = (1.2 - ndre)

    # Moisture scaling
    moisture_adj = np.where(ndmi < 0, 0.7, 1.0)

    # Adjusted Soil Mask based on debug results (BSI mean was 0.67)
    # We apply to pixels where BSI is not extremely high (avoiding buildings/roads)
    soil_mask = (bsi < 0.75)

    # Calculate Prescription
    prescription = base_rate * n_need * moisture_adj * (veg_index > 0.2) * soil_mask

    # Clip and ensure background is 0
    prescription = np.clip(prescription, 0, 220)
    prescription[np.isnan(img[14])] = 0  # Restore original background mask

# Save
meta.update(dtype='float32', count=1, nodata=0)
with rasterio.open(output_rx_path, 'w', **meta) as dst:
    dst.write(prescription.astype(np.float32), 1)

print(f'Updated prescription saved to: {output_rx_path}')
```

### Cell `791daa18`: Calculate Nitrogen Prescription (Initial Attempt)

This cell performs an initial calculation of a variable rate nitrogen prescription map. It reads a stacked GeoTIFF file (`PHMA_VEG_2024_DL.tif`) which is expected to contain various spectral indices.

*   **Input Data**: It opens `PHMA_VEG_2024_DL.tif` using `rasterio`, reading all bands into the `img` array.
*   **Index Extraction**: It extracts specific spectral indices (BSI, EVI, NDMI, NDRE, NDVI, SAVI) from the `img` array using their band indices (10-15). `np.nan_to_num` is used to handle potential NaN values, replacing them with a default value (e.g., 1.0 for BSI, 0.0 for others) to prevent their propagation.
*   **Agronomic Model**: It defines a `base_rate` for nitrogen (120.0 kg/ha) and calculates several factors:
    *   `veg_index`: An average of NDVI, EVI, and SAVI to represent overall vegetation vigor.
    *   `n_need`: Derived from NDRE, where a lower NDRE indicates a higher need for nitrogen (1.2 - NDRE).
    *   `moisture_adj`: Adjusts the rate based on NDMI, assuming less moisture (NDMI < 0) leads to a reduced application rate (0.7 factor).
    *   `soil_mask`: Filters out areas with high BSI (e.g., bare soil, buildings, roads) by applying a mask where `bsi < 0.75`.
*   **Prescription Calculation**: The final `prescription` is calculated by multiplying the `base_rate` with `n_need`, `moisture_adj`, and masks derived from `veg_index` (only apply where `veg_index > 0.2`) and `soil_mask`.
*   **Clipping and Background**: The prescription values are clipped between 0 and 220 kg/ha. Pixels that were originally `NaN` in the NDVI band (indicating non-vegetated or masked areas) are set to 0 in the prescription, effectively restoring the background mask.
*   **Saving Output**: The calculated `prescription` map is saved as a new GeoTIFF file (`Nitrogen_Prescription_Full.tif`) with float32 data type and a `nodata` value of 0. The output confirms the file is saved.

# Variable Rate Nitrogen Prescription Logic
This section calculates the Nitrogen application rates based on spectral indices (NDVI, EVI, NDMI, SAVI, NDRE, BSI) and clips the result to the study area.

```python
import numpy as np
import rasterio
from rasterio.mask import mask
import geopandas as gpd
import matplotlib.pyplot as plt
from matplotlib import cm

# Paths
stacked_path = "/content/drive/MyDrive/NEURAL NETWORK/data/PHMA_VEG_2024_DL.tif"
shapefile_path = "/content/drive/MyDrive/NEURAL NETWORK/data/AKAA/PHMA.shp"
output_rx_path = "/content/drive/MyDrive/NEURAL NETWORK/data/Nitrogen_Prescription_Full.tif"
clipped_rx_path = "/content/drive/MyDrive/NEURAL NETWORK/data/PHMA_Nitrogen_Prescription_Clipped.tif"

# 1. Calculate Prescription
with rasterio.open(stacked_path) as src:
    img = src.read()
    meta = src.meta.copy()

    # Load Indices & Handle NaNs
    bsi  = np.nan_to_num(img[10], nan=1.0)
    evi  = np.nan_to_num(img[11], nan=0.0)
    ndmi = np.nan_to_num(img[12], nan=0.0)
    ndre = np.nan_to_num(img[13], nan=0.0)
    ndvi = np.nan_to_num(img[14], nan=0.0)
    savi = np.nan_to_num(img[15], nan=0.0)

    # Agronomic Model
    base_rate = 120.0
    veg_index = (ndvi + evi + savi) / 3
    n_need = (1.2 - ndre)
    moisture_adj = np.where(ndmi < 0, 0.7, 1.0)
    soil_mask = (bsi < 0.75)

    # Calculate and Mask
    prescription = base_rate * n_need * moisture_adj * (veg_index > 0.2) * soil_mask
    prescription = np.clip(prescription, 0, 220)
    prescription[np.isnan(img[14])] = 0

# 2. Save and Clip
meta.update(dtype="float32", count=1, nodata=0)
with rasterio.open(output_rx_path, "w", **meta) as dst:
    dst.write(prescription.astype(np.float32), 1)

gdf = gpd.read_file(shapefile_path)
with rasterio.open(output_rx_path) as src:
    gdf = gdf.to_crs(src.crs)
    out_image, out_transform = mask(src, gdf.geometry, crop=True)
    out_meta = src.meta.copy()

out_meta.update({"height": out_image.shape[1], "width": out_image.shape[2], "transform": out_transform})
with rasterio.open(clipped_rx_path, "w", **out_meta) as dest:
    dest.write(out_image)

print(f"Clipped prescription saved to: {clipped_rx_path}")
```

Model Architecture - ID CNN

Treat the 16 spectral features as a 1D signal. Conv1D layers to detect local spectral patterns accross adjacent bands and indices

```python
class VegHealthCNN(nn.Module):
    """
    Refined 1D CNN for pixel-based vegetation health classification.
    Optimized for stability and generalization.
    """

    def __init__(self, num_classes=3):
        super(VegHealthCNN, self).__init__()

        self.features = nn.Sequential(
            # Block 1: (1, 16) -> (16, 14)
            # Reduced from 32 to 16 to force more efficient feature extraction
            nn.Conv1d(1, 16, kernel_size=3, padding=0),
            nn.BatchNorm1d(16),
            nn.ReLU(),
            nn.Dropout1d(0.1), # Spatial dropout to reduce jitter

            # Block 2: (16, 14) -> (32, 12)
            nn.Conv1d(16, 32, kernel_size=3, padding=0),
            nn.BatchNorm1d(32),
            nn.ReLU(),
            nn.Dropout1d(0.1),

            # Block 3: (32, 12) -> (64, 10)
            nn.Conv1d(32, 64, kernel_size=3, padding=0),
            nn.BatchNorm1d(64),
            nn.ReLU(),

            # Global average pool: (64, 10) -> (64, 1)
            nn.AdaptiveAvgPool1d(1)
        )

        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Dropout(0.4), # Increased dropout to curb 100% training accuracy
            nn.Linear(64, 32),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(32, num_classes)
        )

        # He Initialization for better convergence
        for m in self.modules():
            if isinstance(m, (nn.Conv1d, nn.Linear)):
                nn.init.kaiming_normal_(m.weight, nonlinearity='relu')

    def forward(self, x):
        # x shape: (batch, 16) -> (batch, 1, 16)
        if x.dim() == 2:
            x = x.unsqueeze(1)
        x = self.features(x)
        x = self.classifier(x)
        return x
```

```python
# Initialize the model
model = VegHealthCNN(num_classes=3)
```

```python
# Print summary
total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"Total parameters:     {total_params:,}")
print(f"Trainable parameters: {trainable_params:,}")
print()
print(model)
```

```python
# Loss, Optimizer, scheduler

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr = 1e-3, weight_decay = 1e-4)
scheduler = ReduceLROnPlateau(optimizer, mode='min', patience=10, factor=0.3)
```

```python
# Training settings
MODEL_SAVE_PATH = '/content/drive/MyDrive/NEURAL NETWORK/MODELS/best_model.pth'
NUM_EPOCHS = 100
PATIENCE = 15
best_val_loss = float('inf')
patience_counter = 0
history = {'train_loss': [], 'val_loss': [], 'train_acc': [], 'val_acc': []}
```

```python
for epoch in range(NUM_EPOCHS):
    # ---- TRAIN ----
    model.train()
    train_loss, train_correct, train_total = 0, 0, 0

    for features, labels in train_loader:

        optimizer.zero_grad()
        outputs = model(features)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        train_loss += loss.item() * features.size(0)
        train_correct += (outputs.argmax(1) == labels).sum().item()
        train_total += features.size(0)

    train_loss /= train_total
    train_acc = train_correct / train_total

    # ---- VALIDATE ----
    model.eval()
    val_loss, val_correct, val_total = 0, 0, 0

    with torch.no_grad():
        for features, labels in test_loader:
            outputs = model(features)
            loss = criterion(outputs, labels)

            val_loss += loss.item() * features.size(0)
            val_correct += (outputs.argmax(1) == labels).sum().item()
            val_total += features.size(0)

    val_loss /= val_total
    val_acc = val_correct / val_total

    # Record history
    history['train_loss'].append(train_loss)
    history['val_loss'].append(val_loss)
    history['train_acc'].append(train_acc)
    history['val_acc'].append(val_acc)

    scheduler.step(val_loss)

    # Print every 5 epochs
    if (epoch + 1) % 5 == 0 or epoch == 0:
        print(f"Epoch {epoch+1:3d}/{NUM_EPOCHS} | "
              f"Train Loss: {train_loss:.4f}  Acc: {train_acc:.4f} | "
              f"Val Loss: {val_loss:.4f}  Acc: {val_acc:.4f}")

    # Early stopping
    if val_loss < best_val_loss:
        best_val_loss = val_loss
        patience_counter = 0
        torch.save(model.state_dict(), MODEL_SAVE_PATH)
    else:
        patience_counter += 1
        if patience_counter >= PATIENCE:
            print(f"\nEarly stopping at epoch {epoch+1}")
            break

print(f"\nBest validation loss: {best_val_loss:.4f}")
print(f"Model saved to: {MODEL_SAVE_PATH}")
```
